```python
# Sample text to learn from
text = "i love natural language processing natural language is fun"

# Step 1: Split the text into a simple list of words
words = text.lower().split()

# Step 2: Count how many times each word appears (Unigram count)
word_counts = {}
for word in words:
    if word in word_counts:
        word_counts[word] += 1
    else:
        word_counts[word] = 1

# Step 3: Count how many times pairs of words appear (Bigram count)
pair_counts = {}
for i in range(len(words) - 1):
    current_word = words[i]
    next_word = words[i + 1]
    
    pair = (current_word, next_word)
    
    if pair in pair_counts:
        pair_counts[pair] += 1
    else:
        pair_counts[pair] = 1

# Step 4: Calculate the probability of a word given the previous word
# Formula: P(next_word | current_word) = Count(pair) / Count(current_word)
probabilities = {}
for pair, count in pair_counts.items():
    first_word = pair[0]
    second_word = pair[1]
    
    prob = count / word_counts[first_word]
    probabilities[f"P({second_word} | {first_word})"] = round(prob, 2)

# --- Print Results ---
print("All Calculated Probabilities:")
for key, value in probabilities.items():
    print(f"{key} = {value}")
```

```python
text = "i love natural language processing natural language is fun"
words = text.split()

# 1. Choose the two words you want to check
word1 = "natural"
word2 = "language"

# 2. Count how many times "natural" appears total
count_word1 = words.count(word1)

# 3. Count how many times "natural language" appears together
count_pair = 0
for i in range(len(words) - 1):
    if words[i] == word1 and words[i+1] == word2:
        count_pair += 1

# 4. Divide the two counts to get probability
probability = count_pair / count_word1

print(f"P('{word2}' after '{word1}') = {probability}")
```
