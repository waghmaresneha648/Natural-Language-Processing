```python
#Extract Email Addresses
import re

text = "Contact us at abc@gmail.com"

emails = re.findall(r'\S+@\S+', text)

print(emails)
```

```python
#Extract Phone Numbers
import re

text = "Call me at 9876543210"

numbers = re.findall(r'\d{10}', text)

print(numbers)
```

```python
#Extract Numbers
import re

text = "I have 5 books and 10 pens."

numbers = re.findall(r'\d+', text)

print(numbers)
```

```python
#Remove Punctuation
import re

text = "Hello!! How are you?"

clean = re.sub(r'[^\w\s]', '', text)

print(clean)
```

```python
#Remove Extra Spaces
import re

text = "Python     is      easy"

clean = re.sub(r'\s+', ' ', text)

print(clean)
```
