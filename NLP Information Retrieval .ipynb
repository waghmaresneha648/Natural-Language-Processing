```python
import sentence_transformers

# Or to import the specific model wrapper directly:
#from sentence_transformers import SentenceTransformer
from  sentence_transformers import SentenceTransformer, util

# 1. Load a fast pre-trained model
model = SentenceTransformer('all-MiniLM-L6-v2')

# 2. Your knowledge base (documents)
documents = [
    "Python is great for machine learning and AI.",
    "Golden retrievers and bulldogs make fantastic house pets.",
    "Natural language processing helps computers read text.",
    "Pizza and pasta are traditional Italian foods."
]

# 3. User query
query = "Tell me about good dogs"

# 4. Embed documents & query, then find highest similarity score
doc_embeddings = model.encode(documents)
query_embedding = model.encode(query)

scores = util.cos_sim(query_embedding, doc_embeddings)[0]
best_match_idx = scores.argmax()

# 5. Output result
print(f"Top Match: {documents[best_match_idx]}")
print(f"Similarity Score: {scores[best_match_idx]:.2f}")
```
