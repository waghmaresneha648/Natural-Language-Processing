```python
#STOP WORD REMOVAL
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

# Download resources (run only once)
nltk.download('punkt')
nltk.download('stopwords')

text = "This is a simple example to demonstrate stop word removal."

# Convert to lowercase
text = text.lower()

# Tokenize
words = word_tokenize(text)

# English stop words
stop_words = set(stopwords.words('english'))

# Remove stop words
filtered_words = []

for word in words:
    if word not in stop_words:
        filtered_words.append(word)

print("Original Words:")
print(words)

print("\nAfter Stop Word Removal:")
print(filtered_words)
```

```python
#TTR (Type Token Ratio)
from nltk.tokenize import word_tokenize
import nltk

nltk.download('punkt')

text = "cat dog cat mouse"

# Tokenize
tokens = word_tokenize(text.lower())

# Count tokens
total_tokens = len(tokens)

# Count unique words
unique_words = set(tokens)

total_types = len(unique_words)

# Calculate TTR
ttr = total_types / total_tokens

print("Tokens:", tokens)
print("Unique Words:", unique_words)
print("Total Tokens:", total_tokens)
print("Total Types:", total_types)
print("Type Token Ratio:", ttr)
```
