```python
import nltk
from nltk.sentiment import SentimentIntensityAnalyzer

# Download the VADER lexicon (only needed once)
nltk.download("vader_lexicon")

# Initialize the analyzer
sia = SentimentIntensityAnalyzer()

# Example sentences
texts = [
    "I absolutely love this product! It works amazingly well.",
    "The service was terrible and the food was cold.",
    "The meeting is scheduled for 3:00 PM today.",
]

print("--- VADER Sentiment Analysis ---")
for text in texts:
    scores = sia.polarity_scores(text)
    print(f"\nText: '{text}'")
    print(f"Positive: {scores['pos']}, Negative: {scores['neg']}, Neutral: {scores['neu']}")
    print(f"Compound Score: {scores['compound']}")
    
    # Classify based on compound score threshold
    if scores['compound'] >= 0.05:
        print("Overall Sentiment: POSITIVE")
    elif scores['compound'] <= -0.05:
        print("Overall Sentiment: NEGATIVE")
    else:
        print("Overall Sentiment: NEUTRAL")
```
