```python
#Regular Expressions in NLP
import re
```

```python
#re.findall

import re

text = "Python is easy. Python is powerful."

result = re.findall("Python", text)

print(result)
```

```python
#re.search()
import re

text = "I love Python."

result = re.search("Python", text)

print(result.group())
```

```python
#re.match()
import re

text = "Python is easy"

result = re.match("Python", text)

print(result.group())
```

```python
#re.sub()
import re

text = "I love Python."

result = re.sub("Python", "Java", text)

print(result)
```

```python
#re.split()
import re

text = "Apple,Orange,Mango"

result = re.split(",", text)

print(result)
```
