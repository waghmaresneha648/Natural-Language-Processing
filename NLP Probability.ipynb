```python
text = input("Enter a sentence: ").lower()
words = text.split()
word1 = input("Enter the first word: ").lower()
word2 = input("Enter the second word: ").lower()
word1_count = words.count(word1)
bigram_count = 0
for i in range(len(words) - 1):
    if words[i] == word1 and words[i + 1] == word2:
        bigram_count = bigram_count + 1
if word1_count == 0:
    print(f"The word '{word1}' does not exist in the text.")
else:
    probability = bigram_count / word1_count
    print(f"Probability of '{word2}' after '{word1}' is: {probability}")
```
