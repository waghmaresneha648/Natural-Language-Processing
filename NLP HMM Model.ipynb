```python
import random



def predict_weather_from_activity(activity):
    """Simple rule-of-thumb guess based on observed activity."""
    if activity == "Cleaning":
        return "Rainy (High chance)"
    elif activity == "Shopping":
        return "Sunny (High chance)"


# --- Let's simulate a 3-day sequence ---
today_weather = "Sunny"  # Start day

print("--- Simulating 3 Days ---")

for day in range(1, 4):
    # Step 1: Pick activity based on today's weather
    if today_weather == "Sunny":
        activity = "Shopping" if random.random() < 0.7 else "Cleaning"
    else:
        activity = "Cleaning" if random.random() < 0.9 else "Shopping"

    # Step 2: HMM guesses the hidden weather just from looking at the activity!
    guessed_weather = predict_weather_from_activity(activity)

    print(
        f"Day {day}: Actual Weather = {today_weather:<6} | Activity Seen = {activity:<8} | HMM Guess = {guessed_weather}"
    )

    # Step 3: Transition to next day's weather
    if today_weather == "Sunny":
        today_weather = "Sunny" if random.random() < 0.8 else "Rainy"
    else:
        today_weather = "Rainy" if random.random() < 0.6 else "Sunny"
```
