```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
#word tokenization
from nltk.tokenize import word_tokenize

text = "Python is an amazing language."
tokens = word_tokenize(text)

print(tokens)
```

```python
#Sentance tokenization
from nltk.tokenize import sent_tokenize

text = "Python is easy. It is popular for AI."

sentences = sent_tokenize(text)

print(sentences)
```

```python
#character tokenization
text = "Python"

tokens = list(text)

print(tokens)
```

```python
#white space tokenization
text = "Python is awesome"

tokens = text.split()

print(tokens)
```

```python
#Regular expression tokenization
from nltk.tokenize import RegexpTokenizer

tokenizer = RegexpTokenizer(r'\w+')

text = "Python, Java & C++ are programming languages."

tokens = tokenizer.tokenize(text)

print(tokens)
```

```python
#Tweet tokenization
from nltk.tokenize import TweetTokenizer

tokenizer = TweetTokenizer()

text = "@john I love #Python 😊"

tokens = tokenizer.tokenize(text)

print(tokens)
```

```python
#Tree tokenization
from nltk.tokenize import TreebankWordTokenizer

tokenizer = TreebankWordTokenizer()

text = "Don't worry, it's fine."

tokens = tokenizer.tokenize(text)

print(tokens)
```

```python
#subword tokenization
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")

tokens = tokenizer.tokenize("unbelievable")

print(tokens)
```

```python
#Byte pair encoding(BPE)
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")

tokens = tokenizer.tokenize("unbelievable")

print(tokens)
```

```python
#
from transformers import GPT2Tokenizer

tokenizer = GPT2Tokenizer.from_pretrained("gpt2")

tokens = tokenizer.tokenize("tokenization")

print(tokens)
```

```python
#word pice tokenization
from transformers import BertTokenizer

tokenizer = BertTokenizer.from_pretrained("bert-base-uncased")

tokens = tokenizer.tokenize("playing")

print(tokens)
```

```python
#Byte level tokenization
from transformers import GPT2Tokenizer

tokenizer = GPT2Tokenizer.from_pretrained("gpt2")

encoded = tokenizer.encode("Hello!")

print(encoded)
```
