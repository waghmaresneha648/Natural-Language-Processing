```python
import spacy

# Load the small English pipeline model
try:
    nlp = spacy.load("en_core_web_sm")
except OSError:
    print("Model not found. Run: python -m spacy download en_core_web_sm")
    exit()

def run_ner():
    print("=== Named Entity Recognition (NER) Interactive Tool ===")
    print("Type 'exit' or 'quit' to stop.\n")
    
    while True:
        # 1. Get user input
        user_input = input("Enter text: ").strip()
        
        if user_input.lower() in ['exit', 'quit']:
            print("Exiting...")
            break
            
        if not user_input:
            continue

        # 2. Process text through the NLP pipeline
        doc = nlp(user_input)

        # 3. Check and display entities
        if not doc.ents:
            print("\n  -> No named entities found in this text.\n")
        else:
            print(f"\n  Found {len(doc.ents)} entity/entities:")
            print("  " + "-" * 50)
            print(f"  {'Entity Text':<25} | {'Type':<10} | {'Explanation'}")
            print("  " + "-" * 50)
            
            for ent in doc.ents:
                explanation = spacy.explain(ent.label_)
                print(f"  {ent.text:<25} | {ent.label_:<10} | {explanation}")
            print("  " + "-" * 50 + "\n")

if __name__ == "__main__":
    run_ner()
```
